```python
import pandas as pd 
root_path = "C:\\Users\\shibiyasaju\\Desktop\\DATA ANALYTICS\\main case study\\Road Accident Analytics and Road Safety Optimization in India"
df = pd.read_csv(f"{root_path}\\data\\csv\\accident_prediction_india.csv")

df.head()

```

```python
df.columns
```

```python
df["Number of Casualties"].describe()
```

```python
severity_stats = (
    df.groupby("Accident Severity")["Number of Casualties"]
      .agg(["count", "mean", "median", "std", "min", "max"])
)

severity_stats
```

```python
road_stats = (
    df.groupby("Road Type")["Number of Casualties"]
      .agg(["count", "mean", "median", "std", "min", "max"])
      .sort_values("mean", ascending=False)
)

road_stats
```

```python
weather_stats = (
    df.groupby("Weather Conditions")["Number of Casualties"]
      .agg(["count", "mean", "median", "std"])
      .sort_values("mean", ascending=False)
)

weather_stats
```

```python
vehicle_stats = (
    df.groupby("Vehicle Type Involved")["Number of Casualties"]
      .agg(["count", "mean", "median", "std"])
      .sort_values("mean", ascending=False)
)

vehicle_stats
```

```python
severity_percentage = (
    df["Accident Severity"]
    .value_counts(normalize=True) * 100
)

severity_percentage
```

```python
road_percentage = (
    df["Road Type"]
    .value_counts(normalize=True) * 100
)

road_percentage
```

```python
fatal = (df["Accident Severity"] == "Fatal").sum()

minor = (df["Accident Severity"] == "Minor").sum()

fatal_to_minor_ratio = fatal / minor

fatal_to_minor_ratio
```

```python
yearly_accidents = (
    df.groupby("Year")
      .size()
      .reset_index(name="Accidents")
)

yearly_accidents
```

```python
yearly_accidents["YoY Growth %"] = (
    yearly_accidents["Accidents"]
    .pct_change() * 100
)

yearly_accidents
```

```python
highest_year = yearly_accidents.loc[
    yearly_accidents["Accidents"].idxmax()
]

highest_year
```

```python
severity_year = (
    pd.crosstab(
        df["Year"],
        df["Accident Severity"],
        normalize="index"
    ) * 100
)

severity_year
```

```python
state_casualties = (
    df.groupby("State Name")["Number of Casualties"]
      .sum()
      .sort_values(ascending=False)
)

state_casualties
```

```python
df["Fatal_Flag"] = (
    df["Accident Severity"] == "Fatal"
).astype(int)
```

```python
weather_fatality = (
    df.groupby("Weather Conditions")["Fatal_Flag"]
      .mean() * 100
)

weather_fatality.sort_values(ascending=False)
```
