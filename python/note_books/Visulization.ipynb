```python
fatal_by_vehicle = (
    df.assign(
        Is_Fatal=df['Accident Severity']
        .astype('string')
        .str.strip()
        .str.casefold()
        .eq('fatal')
    )
    .groupby('Vehicle Type Involved', dropna=False)
    ['Is_Fatal']
    .agg(['sum', 'count', 'mean'])
    .reset_index()
)

fatal_by_vehicle['Fatal Percentage'] = (
    fatal_by_vehicle['mean'] * 100
)

fatal_by_vehicle = fatal_by_vehicle.sort_values(
    'Fatal Percentage',
    ascending=False
)

print(fatal_by_vehicle)
```

```python
year_counts = (
    df.groupby('Year')
      .size()
      .reset_index(name='Accident Count')
      .sort_values('Year')
)

plt.figure(figsize=(9, 5))

sns.lineplot(
    data=year_counts,
    x='Year',
    y='Accident Count',
    marker='o'
)

plt.title('Yearly Trend in Recorded Accidents')
plt.xlabel('Year')
plt.ylabel('Number of Records')
plt.xticks(year_counts['Year'])
plt.tight_layout()
plt.show()
```

```python
year_counts['Previous Year Count'] = (
    year_counts['Accident Count'].shift(1)
)

year_counts['Change in Count'] = (
    year_counts['Accident Count']
    - year_counts['Previous Year Count']
)

year_counts['Change Percentage'] = (
    year_counts['Change in Count']
    .div(year_counts['Previous Year Count'])
    .mul(100)
)

print(year_counts.round(2))
```

```python
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_counts = (
    df['Month']
    .value_counts()
    .reindex(month_order, fill_value=0)
    .rename_axis('Month')
    .reset_index(name='Accident Count')
)

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=monthly_counts,
    x='Month',
    y='Accident Count',
    marker='o'
)

plt.title('Recorded Accidents by Month')
plt.xlabel('Month')
plt.ylabel('Number of Records')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

day_counts = (
    df['Day of Week']
    .value_counts()
    .reindex(day_order, fill_value=0)
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=day_counts.index,
    y=day_counts.values
)

plt.title('Recorded Accidents by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Number of Records')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
heatmap_data = pd.crosstab(
    df['Month'],
    df['Day of Week']
)

heatmap_data = heatmap_data.reindex(month_order)

heatmap_data = heatmap_data.reindex(
    columns=day_order
)

plt.figure(figsize=(11, 7))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='d',
    cmap='YlOrRd'
)

plt.title('Accident Counts by Month and Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Month')
plt.tight_layout()
plt.show()
```

```python
numeric_data = df[
    [
        'Number of Vehicles Involved',
        'Number of Casualties',
        'Number of Fatalities',
        'Speed Limit (km/h)',
        'Driver Age'
    ]
]

correlation_matrix = numeric_data.corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    center=0
)

plt.title('Correlation Between Numerical Variables')
plt.tight_layout()
plt.show()
```
